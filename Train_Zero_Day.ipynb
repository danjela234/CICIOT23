1. Load data

```python
import pandas as pd

df = pd.read_csv("train/train.csv")
```

2. Create target (binary)

```python
df["y"] = df["label"].apply(lambda x: 0 if x == "BenignTraffic" else 1)
```

3. Prepare features

```python
X = df.drop(columns=["label", "y"], errors="ignore")
y = df["y"]

# numeric only
X = X.apply(pd.to_numeric, errors="coerce").fillna(0)
```

4. Remove useless features

```python
X = X.loc[:, X.nunique() > 1]
```

5. Train/test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

6. Train model

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

7. Evaluate

```python
from sklearn.metrics import classification_report

y_pred = rf.predict(X_test)

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Benign', 'Attack'],
            yticklabels=['Benign', 'Attack'])

plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix - Zero-Day')
plt.tight_layout()

plt.savefig('cm_zeroday.png', dpi=300)
plt.show()
```

8. SHAP

```python
import shap

# small sample
X_sample = X_test.sample(100, random_state=42)

explainer = shap.TreeExplainer(rf)

shap_values = explainer(X_sample)

# select ONE class (attack class)
shap.plots.beeswarm(shap_values[:, :, 1])
```
