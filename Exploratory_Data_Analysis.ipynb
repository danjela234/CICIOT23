1. Counting total rows of train

```python
import pandas as pd

TRAIN_PATH = r"train\train.csv"

total_rows = 0

for chunk in pd.read_csv(TRAIN_PATH, chunksize=500000):
    total_rows += len(chunk)

print("Total rows in train.csv:", total_rows)
```

2. Counting attack distribution

```python
from collections import Counter
import pandas as pd

TRAIN_PATH = r"train\train.csv"

label_counts = Counter()

for chunk in pd.read_csv(TRAIN_PATH, chunksize=500000):
    label_counts.update(chunk["label"])

print("Number of unique attack types:", len(label_counts))

print("\nTop 20 most common labels:\n")
for label, count in label_counts.most_common(20):
    print(label, ":", count)
```
